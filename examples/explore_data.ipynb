# Explore HD5 data-stores

This notebook lists the hd5 datastructure of all files in the `data` folder.

```python
import os
import pandas as pd

data_folder = "../data"
print("Data folder: ", os.path.abspath(data_folder))
# Find all .hd5 and .hdf files in the data folder
hd5_files = []
for root, dirs, files in os.walk(data_folder):
    for f in files:
        if f.lower().endswith((".hd5", ".hdf")):
            hd5_files.append(os.path.join(root, f))

file_keys = {}

for fpath in hd5_files:
    try:
        # Check if this is a valid pandas HDF store
        with pd.HDFStore(fpath, mode='r') as store:
            # Only collect if it is a pandas-format store
            if store.is_open:
                keys = []
                try:
                    keys = store.keys()
                except Exception as e_keys:
                    print(f"  Error reading keys in {fpath}: {e_keys}")
                    keys = []
                # Optionally: skip if no keys
                file_keys[fpath] = keys
    except (OSError, IOError, ValueError) as e:
        # OSError: not an HDF5 file, or some HDF5 error
        # ValueError: underlying format error
        print(f"Error opening (non-pandas?) {fpath}: {e}")
    except Exception as e:
        # Catch-all for any other exceptions
        print(f"General error accessing {fpath}: {e}")

for fname, keys in file_keys.items():
    print(f"\nFile: {fname}")
    if not keys:
        print("  (No pandas-format datasets found)")
    for k in keys:
        print(f"  {k}")
```
